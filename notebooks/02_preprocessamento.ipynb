## 2. Definição da variável alvo

A variável alvo foi consolidada em formato binário: 0 para Bons Pagadores e 1 para Maus Pagadores. O limiar de corte escolhido foi a presença de pelo menos um histórico de atraso igual ou superior a 60 dias, conforme descrito no notebook 1.
A escolha foi feita considerando que pequenos atrasos são volateis e costumam ser regularizados brevemente e atrasos maiores possuem maior risco de não chegarem a ser pagos em algum momento.
